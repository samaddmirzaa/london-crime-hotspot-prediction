### Data Cleaning

#### Importing Packages and Loading Data

```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
# Setting up Project Paths
project_root = Path.cwd().parent
raw_dir = project_root/"data"/"raw"
processed_dir = project_root/"data"/"processed"
```

```python
# Loading Data Frame
frames = []
for f in sorted(raw_dir.rglob("*.csv")):
    df_part = pd.read_csv(f)
    df_part["source_file"] = f.name
    frames.append(df_part)

df = pd.concat(frames, ignore_index=True)

df
```

#### Dropping Columns and Rows

```python
# "Context" is fully empty
df = df.drop(columns=["Context"])
```

```python
df.head()
```

```python
df.shape
```

```python
# Dropping rows with no Locations
df = df.dropna(subset=["LSOA code", "Longitude", "Latitude"])
df.shape
```

#### Changing Column Names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.head()
```

#### Removing out of London crimes

```python
# The Metroploitan Police sometimes reports crimes which happen just outside the London boundary. Greater London sits roughly within longitude −0.52 to 0.33 and latitude 51.28 to 51.70. So we keep everything inside this boundary.

lon_min, lon_max = -0.52, 0.33
lat_min, lat_max = 51.28, 51.70

in_london = (
    df["longitude"].between(lon_min, lon_max)
    & df["latitude"].between(lat_min, lat_max)
)
df.shape
```

```python
df = df[in_london]
df.shape
```

#### Duplicate Rows

```python
df.duplicated().value_counts()
```

```python
df[df.duplicated()]
```

```python
dup_mask = df.duplicated(keep="first")
```

```python
df.loc[dup_mask, "crime_type"].value_counts()
```

##### Anti-social behavior has no crime ID and has only month level dates. Separate incidents of the same kind are being flagged as duplicate but are real separate incidents and we shall keep them in our data. We will remove the duplicates that have a crime ID.

```python
has_cid = df["crime_id"].notna()
is_dup = df.duplicated(keep="first")

df.shape
```

```python
dropping = has_cid & is_dup
df = df[~dropping]

df.shape
```

```python
df
```

```python
# Dropping Falls Within column
df = df.drop(columns=["falls_within"])

df
```

#### Saving the Dataframe

```python
print("Shape:", df.shape)
print("\nRemaining missing values:")
print(df.isna().sum())
```

```python
df.to_parquet(processed_dir/"london_crime.parquet", index=False)
```
