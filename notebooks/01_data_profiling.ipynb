### Data Profiling

#### Importing Packages and Data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
# Setting up Project Paths

project_root = Path.cwd().parent
raw_dir = project_root/"data"/"raw"
csv_files = sorted(raw_dir.rglob("*.csv"))
```

```python
# Setting up Data Frame

frames = []

for f in csv_files:
    df_part = pd.read_csv(f)
    df_part["source_file"] = f.name
    frames.append(df_part)

df = pd.concat(frames, ignore_index=True)
```

```python
# Missing Values
missing = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_pct": (df.isna().mean() * 100).round(2)
})
print(missing)
```

```python
# Are all 24 months present, and how many crimes per month?
print(df["Month"].value_counts().sort_index())

# Which Police force reported it in the data?
print(df["Reported by"].value_counts())

# What crime types exist, and are there exactly 14?
print(f"\nCrime types: {df['Crime type'].nunique()}")
print(df["Crime type"].value_counts())

# Are there exact duplicated rows?
dupes = df.duplicated().sum()
print(f"\n Duplicate rows: {dupes:,}")
```
